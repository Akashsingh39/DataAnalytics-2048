### 2. `Flatten() and ravel()`

Both Converts a multidimensional array into one-dimension, but their memeory behavior differ.

```python
arr = np.array([
    [1,2],
    [3,4]
])
print(arr)
```

```python
one_d =arr.flatten()
print(one_d)
```

```python
one_d1 = arr.ravel()
print(one_d1)
```

```python
## 2-D to 1-D

arr1 = np.arange(1,13)

arr = arr1.reshape(3,4)
print(arr)
matrix = arr.flatten()
print(matrix)

matrix1 = arr.ravel()
print(matrix1)
```

```python
# 3-D to 1- D 
arr1 = np.arange(1,13)

arr = arr1.reshape(2,2,3)
print(arr)
matrix = arr.flatten()
print(matrix)

matrix1 = arr.ravel()
print(matrix1)
```

```python
matrix1
```

Notes:

flatten(): return a copy.

ravel(): returna view when possible other a copy.

```python
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr)
```

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60]])

result = arr.flatten()

print(result)
# [10 20 30 40 50 60]

result[0] = 100

print("Original:", arr)
print("Flatten:", result)
```

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60]])

result = arr.ravel()

print(result)
# [10 20 30 40 50 60]

result[0] = 100

print("Original:", arr)
print("Ravel:", result)

result[1] = 500
print(result)
```

### 3. Transpose 


Transpose swaps the axis of an array

```python
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr)
print(arr.shape)
```

```python
arrT = arr.T
print(arrT)
print(arrT.shape)
```

Note: For a two-dimensional array. `.T` is best way to exchange rows and columns.
    

```python
arr = np.arange(1,25)
print(arr)
```

```python
matrix = arr.reshape(2,2,6)
print(matrix)
print(matrix.shape)
```

```python
matrixT = matrix.T
print(matrixT)
print(matrixT.shape)
```

### 4. `Squeeze() and expand_dims()`

`squeeze()` removs axes of length 1

```python
arr = np.array([[[10,20,30]]])
print(arr)
```

```python
print(arr.shape)
```

```python
print(arr.ndim)
```

```python
print(np.squeeze(arr))
print(np.squeeze(arr).shape)
print(np.squeeze(arr).ndim)
```

`expand_dims()` adds a dimension at a specific posisiton

```python
arr = np.array([10,20,30])
print(arr)

print(arr.shape) # 3 columns

print(np.expand_dims(arr,axis=0))  # converted into rows and columns - 2-d
print(np.expand_dims(arr,axis=0).shape)
print(np.expand_dims(arr,axis=1))  # converted into 3rows ad 1 columns
print(np.expand_dims(arr,axis=1).shape)
```

### Notes:

These operations are particulary useful when matching shapes for machine-learning models broadcast.
