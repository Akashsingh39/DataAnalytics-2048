#### Vectorization

Vectorization means applying an operation to an entire array instead of looping through it's element in python.

without Numpy

````python
numbers = [1,2,3,4]

squares = []

for i in numbers:
    squares.append(i**2)
print(squares)


arr = np.array(numbers)
print(arr)
````

````python
arrSqr = []
for i in arr:
    arrSqr.append(i**2)
arrSqr
````

````python
numbers = [1,2,3,4]
print(numbers)

arr = np.array(numbers)
print(arr)


arrSqr = arr**2
print(arrSqr)
````
