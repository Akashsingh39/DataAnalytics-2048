## Syntax of np.where()

```python
np.where(condition, value_if_true, value_if_false)
```

## Section A: 1-D Array (Questions 1–7)

**Q1.** Find all elements greater than 50 and replace them with 100. Replace remaining elements with 0.

```python
import numpy as np
arr = np.array([10, 60, 30, 80, 45, 90])
```


````python
import numpy as np
arr = np.array([10, 60, 30, 80, 45, 90])

result = np.where(arr>50,100,0)
print(result)
````

````python
# Replace all even numbers with 1 and odd numbers with 0. --> 

arr = np.array([12, 15, 20, 23, 30, 35, 40])

result = np.where(arr%2 == 0, 1,0)
print(result)
````

````python
# Q3. Identify students who scored 40 or above. Return "Pass" otherwise "Fail".

marks = np.array([35, 75, 28, 90, 45, 32, 60])

result = np.where(marks>=40,"Pass","Fail")
print(result)
````

````python
# Q4. Replace negative numbers with 0 and keep positive numbers unchanged.

arr = np.array([-10, 20, -5, 30, -15, 40])

result = np.where(arr<0, 0, arr)
print(result)
````

````python
# Q5. Replace numbers divisible by 5 with 500; otherwise keep the original number.

arr = np.array([10, 12, 15, 23, 25, 31, 40])

result = np.where(arr%5==0,500,arr)

print(result)
````

````python
# Q6. Identify numbers between 20 and 50 (inclusive). Return "Yes" or "No".

arr = np.array([10, 25, 45, 60, 20, 50, 15])

result = np.where((arr>=20) & (arr<=50), "Yes","No")
print(result)
````

````python
# Q7. Replace marks below 33 with 0, marks from 33 to 59 with 1, and marks 60 or above with 2. Use nested np.where().

marks = np.array([25, 45, 70, 32, 55, 90, 60])

result = np.where(marks<33,0,
                   np.where(marks>=60,1,2)
                  )
print(result)
````

## Section B: 2-D Array (Questions 8–14)


````python
# Q8. Replace all elements greater than 50 with 999.

arr = np.array([
    [10, 60, 30],
    [80, 45, 90],
    [20, 70, 40]
])

result = np.where(arr>50,999,arr)
print(result)
````

````python
# Q9. Replace all even numbers with 0 and odd numbers with 1.

arr = np.array([
    [12, 15, 20],
    [23, 30, 35],
    [40, 45, 50]
])

result = np.where(arr%2==0,0,1)
print(result)
````

````python
# Q10. Find all marks below 40 and replace them with "Fail", otherwise "Pass".

marks = np.array([
    [35, 75, 28],
    [90, 45, 32],
    [60, 25, 85]
])

result = np.where(marks<40,"Fail","Pass")
print(result)
````

````python
# Q11. Replace negative values with 0 and positive values with 1.

arr = np.array([
    [-10, 20, -5],
    [30, -15, 40],
    [-25, 50, -35]
])

result = np.where(arr<0, 0, 1)
print(result)
````

````python
# Q12. Replace all numbers divisible by 3 with 100.

arr = np.array([
    [9, 10, 12],
    [14, 15, 17],
    [18, 20, 21]
])

result = np.where(arr%3==0,100,arr)
print(result)
````

````python
# Q13. Find values greater than the average of the complete array. Return 1 for greater values and 0 otherwise.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

avg = np.mean(arr)
print(avg)

result = np.where(arr>avg,1,0)
print(result)
````

````python
# Q14. Replace marks according to these conditions:

# Below 33 → "Fail"
# 33 to 59 → "Average"
# 60 or above → "Good"

marks = np.array([
    [25, 45, 70],
    [32, 55, 90],
    [60, 20, 85]
])

result = np.where(marks<33,"Fail",
                 np.where(marks>=60,"Good","Average"))

print(result)
````

## Section C: 3-D Array (Questions 15–20)


````python
# Q15. Replace all elements greater than 50 with 999.

arr = np.array([
    [[10, 60], [30, 80]],
    [[45, 90], [20, 70]]
])

result = np.where(arr>50,999,arr)
print(result)
````

````python
# Q16. Replace even numbers with 0 and odd numbers with 1.

arr = np.array([
    [[12, 15], [20, 23]],
    [[30, 35], [40, 45]]
])

result = np.where(arr%2==0,0,1)
print(result)
````

````python
# Q17. Replace negative numbers with 0 and positive numbers with their original values.

arr = np.array([
    [[-10, 20], [-5, 30]],
    [[-15, 40], [25, -35]]
])

result = np.where(arr<0,0,arr)
print(result)
````

````python
# Q18. Find all elements divisible by 5. Replace them with 500; otherwise replace them with -1.

arr = np.array([
    [[10, 12], [15, 23]],
    [[25, 31], [40, 42]]
])

result = np.where(arr%5==0,500,-1)
print(result)
````

````python
# Q19. Identify elements greater than 50. Return "High" otherwise "Low".

arr = np.array([
    [[10, 60], [30, 80]],
    [[45, 90], [20, 70]]
])

result = np.where(arr>50,"High","Low")
print(result)
````

````python
# Q20. Apply three conditions using nested np.where():

# Below 30 → 0
# 30 to 60 → 1
# Above 60 → 2

arr = np.array([
    [[10, 35], [60, 80]],
    [[25, 45], [70, 90]]
])

result = np.where(arr<30,0,
                 np.where(arr>60,2,1))

print(result)
````

## Quick Revision: NumPy Operators

| Operator | Meaning |
|---|---|
| `>` | Greater than |
| `<` | Less than |
| `==` | Equal to |
| `!=` | Not equal to |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |
| `&` | Both conditions (AND) |
| `\|` | Either condition (OR) |


````python
arr = np.array([
    [[10, 35], [60, 80]],
    [[25, 45], [70, 90]]
])

result = np.where(arr<30,0,
                 np.where(arr<60,1,
                 np.where(arr<=90,2,3)))

print(result)
````
