```python
# Q1. Add 10 to every element of the following NumPy array using broadcasting.
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print(arr+10)
```

```python
# Q7. Calculate the percentage of each student using broadcasting.

marks = np.array([[80, 70, 90],
                  [60, 75, 85],
                  [90, 95, 80]])

total = np.array([100, 100, 100])

percentage = (marks/total)*100
print(percentage)
```

```python
# Q11. Add two arrays of shapes (3, 1) and (1, 3). Find the resulting shape.
arr1 = np.ones((3, 1))
arr2 = np.ones((1, 3))

result = arr1 + arr2

print(result)
print("Shape:", result.shape)
```

```python
# Q12. Subtract the column-wise minimum values from a matrix using broadcasting.

arr = np.array([
    [10,3,30],
    [2,4,1],
    [4,5,6]
])

col_min = np.min(arr,axis=0)

print(col_min)
print(arr - col_min)
```

```python
# Q16. Calculate the square of every element without using loops.

arr = np.array([2, 4, 6, 8, 10])

result = arr**2
print(result)
```

```python
# Q17. Calculate the cube of every element using vectorization.

arr = np.array([1, 2, 3, 4, 5])

print(arr**3)
```

```python
# Q18. Extract all even and odd numbers from the following array.

arr = np.array([12, 15, 18, 21, 24, 27, 30])

even = arr[arr%2 == 0]
odd = arr[arr%2 != 0]

print("Even : ",even)
print("Odd : ",odd)
```

```python
# Q19. Replace all marks below 40 with 0 and marks greater than or equal to 40 with 1.

marks = np.array([35, 45, 78, 90, 28, 65, 88])

result = np.where(marks < 40,0,1)
print(result)
```

```python
# Q20. Calculate a 10% discount on every product price.

prices = np.array([1000, 2000, 3000, 4000, 5000])

discount = prices * 10 / 100
print(discount)
```

```python
# Q21. Calculate 5% tax on every employee's salary.

salary = np.array([20000, 30000, 40000, 50000, 60000])

tax = salary * 5 /100
print(tax)
```

```python
# Q22. Find the maximum, minimum, difference and index of the maximum value.

arr = np.array([45, 12, 78, 34, 90, 23, 67])

maximum = np.max(arr)
minimum = np.min(arr)
difference = maximum - minimum
index_max = np.argmax(arr)

print(maximum)
print(minimum)
print(difference)
print(index_max)
```

```python
# Q23. Calculate the Euclidean distance between two arrays using vectorization.
import numpy as np

x = np.array([2, 4, 6, 8])
y = np.array([1, 3, 5, 7])

distance = np.sqrt(np.sum((x - y) ** 2))

print("Euclidean Distance:", distance)
```

```python
# Q24. Replace all negative values with zero without using loops.

arr = np.array([-10, 20, -30, 40, -50, 60])

result = np.where(arr<0,0,arr)
print(result)
```

```python
# Q25. Calculate the total and average marks of every student.

marks = np.array([[80, 75, 90],
                  [35, 45, 50],
                  [90, 85, 95],
                  [25, 30, 40],
                  [60, 70, 80]])

total = np.sum(marks,axis=1)
avg = np.mean(marks,axis=1)

print("Total Marks: ",total)
print("Average Marks: ",avg)
```

```python
# Q26. Find the highest and lowest marks in each subject.
marks = np.array([[80, 75, 90],
                  [35, 45, 50],
                  [90, 85, 95],
                  [25, 30, 40],
                  [60, 70, 80]])

highest = np.max(marks,axis=0)
lowest = np.min(marks,axis=0)

print("highest marks: ",highest)
print("lowest marks: ",lowest)
```

```python
# Q27. Count the number of students whose average marks are greater than or equal to 40.
marks = np.array([[80, 75, 90],
                  [35, 45, 50],
                  [90, 85, 95],
                  [25, 30, 40],
                  [60, 70, 80]])

avg = np.mean(marks, axis=1)
count = np.sum(avg>=40)

print("Average marks: ",avg)
print("count of average marks greater than or equal to 40: ",count)
```

```python
# Q30. Normalize every element of the array between 0 and 1.
import numpy as np

arr = np.array([10,20,30,40,50])

result = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

print(result)
```

```python
# Q31. Apply different discount percentages to each product column using broadcasting.

sales = np.array([[1000, 2000, 3000],
                  [1500, 2500, 3500],
                  [2000, 3000, 4000]])

discount = np.array([10, 15, 20])

final_price = sales - (sales*discount/100)
print(final_price)
```

```python
# Q32. Convert the following Celsius temperatures into Fahrenheit using vectorization.

celsius = np.array([[20, 25, 30],
                    [15, 22, 28],
                    [10, 18, 24]])

Fahrenheit = (celsius * 9/5) + 32

print(Fahrenheit)
```

```python
# Q33. Normalize each column using Min-Max normalization.

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

col_min = np.min(arr, axis=0, keepdims=True)
col_max = np.max(arr, axis=0, keepdims=True)

result = (arr - col_min) / (col_max - col_min)
print(result)
```

```python
# Q34. Calculate the total salary after adding bonuses and find employees whose total salary is above 40,000.

salary = np.array([25000, 35000, 45000, 55000, 65000])
bonus = np.array([2000, 3000, 4000, 5000, 6000])

total_salary = salary + bonus

result = total_salary[total_salary > 40000]

print(result)
```

```python
# Q35. Calculate the Euclidean distance of every point from the reference point using broadcasting and vectorization.

points = np.array([[2, 3],
                   [4, 5],
                   [6, 7],
                   [8, 9]])

reference = np.array([1, 1])

result = np.sqrt(np.sum((points - reference) ** 2, axis=1))
print(result)
```
