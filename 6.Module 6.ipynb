Joining and Spliting array.

```python
import numpy as np
```

```python
a = np.array([1,2,3])  # 1-D
print(a)

b = np.array([4,5,6])  # 1-d
print(b)
```

```python
print(np.concatenate((a,b)))  # 1-D
```

```python
a = np.array([
    [1,2],
    [3,4]
])
print(a)

b = np.array([
    [4,5],
    [6,7]
])
print(b)
```

```python
print(np.concatenate((a,b)))
```

```python
print(np.concatenate((a,b),axis=0))
```

```python
print(np.concatenate((a,b),axis=1))
```

```python
arr = np.arange(12)
print(arr)

matrix1 = arr.reshape(2,2,3)
print(matrix1)
arr1 = np.arange(13,25)
print(arr1)
matrix2 = arr1.reshape(2,2,3)
print(matrix2)
```

```python
print(np.concatenate((matrix1,matrix2)))
```

```python
print(np.concatenate((matrix1,matrix2),axis=0))
```

```python
print(np.concatenate((matrix1,matrix2),axis=1))
```

### `stack(), hstack() and vstack()`

stack() joins array along a new axis

```python
import numpy as np
a = np.array([1,2])
b = np.array([3,4])
print(a)
print(b)
```

```python
np.stack((a,b)) # rows and columns
```

```python
np.stack((a,b),axis=0) # rows and columns
```

Horizontal and Vertical



```python
import numpy as np
a = np.array([1,2])
b = np.array([3,4])
print(a)
print(b)
```

```python
np.hstack((a,b))
```

```python
np.vstack((a,b))
```

### 25. Splitting array
