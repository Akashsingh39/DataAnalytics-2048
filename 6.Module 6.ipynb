Joining and Spliting array.

```python
import numpy as np
```

```python
a = np.array([1,2,3])  # 1-D
print(a)

b = np.array([4,5,6])  # 1-d
print(b)
```

```python
print(np.concatenate((a,b)))  # 1-D
```

```python
a = np.array([
    [1,2],
    [3,4]
])
print(a)

b = np.array([
    [4,5],
    [6,7]
])
print(b)
```

```python
print(np.concatenate((a,b)))
```

```python
print(np.concatenate((a,b),axis=0))
```

```python
print(np.concatenate((a,b),axis=1))
```

```python
arr = np.arange(12)
print(arr)

matrix1 = arr.reshape(2,2,3)
print(matrix1)
arr1 = np.arange(13,25)
print(arr1)
matrix2 = arr1.reshape(2,2,3)
print(matrix2)
```

```python
print(np.concatenate((matrix1,matrix2)))
```

```python
print(np.concatenate((matrix1,matrix2),axis=0))
```

```python
print(np.concatenate((matrix1,matrix2),axis=1))
```
